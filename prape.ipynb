```python
# ── Install all dependencies (CLEAN + COMPLETE) ─────────────────────────

import subprocess, sys

PACKAGES = [
    # Core
    "openai",
    "python-dotenv",
    "requests",
    "tiktoken",

    # LangChain ecosystem
    "langchain",
    "langchain-core",
    "langchain-community",
    "langchain-openai",
    "langchain-text-splitters",

    # Agents / workflows
    "langgraph",

    # Search / tools
    "duckduckgo-search",
    "wikipedia",
    "tavily-python",

    # Math / evaluation
    "numexpr",

    # Data / analysis (used later or useful)
    "pandas",
    "numpy",
    "matplotlib",
    "seaborn",

    # Vector DBs / embeddings
    "sentence-transformers",
    "chromadb",
    "faiss-cpu",
    "qdrant-client",

    # Document parsing
    "pypdf",
    "unstructured",

    # API serving (optional but included)
    "fastapi",
    "uvicorn",

    # Notebook environment
    "ipython",
    "notebook",
    "jupyterlab",
    "ipykernel",

    # Observability
    "langsmith"
]

print("📦 Installing dependencies...\n")

failed = []

for pkg in PACKAGES:
    try:
        subprocess.check_call(
            [sys.executable, "-m", "pip", "install", pkg, "-q", "--upgrade"]
        )
        print(f"✅ {pkg}")
    except subprocess.CalledProcessError:
        print(f"❌ {pkg}")
        failed.append(pkg)

print("\n🎉 Installation complete!")

if failed:
    print(f"\n⚠️ Failed packages: {failed}")
else:
    print("\n✅ All packages installed successfully!")
```
